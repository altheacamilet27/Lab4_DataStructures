```python
# ================================
# EXERCISE 1: UNSTRUCTURED DATA AGGREGATOR
# ================================

from collections import defaultdict, Counter

stream = [
    ("TUP Manila", "BSECE"),
    ("TUP Taguig", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", "BSECE"),
    ("TUP Visayas", STUDENT_MAJOR),
    ("TUP Taguig", "BSECE"),
    ("TUP Manila", "BSECE")
]

for i in range(CONTROL_NUM + 3):
    stream.append(("TUP Manila", f"CUSTOM_{i}"))

grouped = defaultdict(list)

for campus, program in stream:
    grouped[campus].append(program)

manila_counter = Counter(grouped["TUP Manila"])
top_program = manila_counter.most_common(1)[0]

print("STREAM LENGTH:", len(stream))
print("MANILA TOTAL:", len(grouped["TUP Manila"]))
print("TOP PROGRAM:", top_program[0])
print("FREQUENCY:", top_program[1])
```

```python
# ================================
# EXERCISE 2: JACCARD SIMILARITY ENGINE
# ================================

festival = {
    "Ben&Ben", "SB19", "Bini", "Eraserheads",
    FAVORITE_ARTIST, "Zild", f"Indie Artist {CONTROL_NUM}"
}

user_a = {"Ben&Ben", "Bini", "Maki", "Dionela", FAVORITE_ARTIST}
user_b = {"SB19", "Eraserheads", "Zild", f"Indie Artist {CONTROL_NUM}", "Parokya ni Edgar"}

# add CONTROL_NUM new artists
for i in range(CONTROL_NUM):
    festival.update({f"NEW_ARTIST_{i}"})

# INTERSECTION & UNION
inter_a = user_a & festival
union_a = user_a | festival

inter_b = user_b & festival
union_b = user_b | festival

# JACCARD FORMULA
jac_a = (len(inter_a) / len(union_a)) * 100
jac_b = (len(inter_b) / len(union_b)) * 100

# DEALBREAKERS
dealbreakers = user_a - festival

# ================================
# OUTPUTS (REQUIRED TABLE VALUES)
# ================================

print("TOTAL ARTISTS IN FESTIVAL:", len(festival))
print("INTERSECTION (USER A & FESTIVAL):", len(inter_a))
print("UNION (USER A & FESTIVAL):", len(union_a))
print("USER A JACCARD (%):", jac_a)
print("USER B JACCARD (%):", jac_b)
print("USER A DEALBREAKERS:", dealbreakers)
```

```python
# ================================
# EXERCISE 3: SLIDING WINDOW FILTER SYSTEM
# ================================

from collections import deque, Counter

# initialize deque with memory limit
dq = deque(maxlen=5)

# base burst
burst = [
    (CONTROL_NUM, "WEATHER"),
    (CONTROL_NUM + 1, "TRAFFIC"),
    (CONTROL_NUM + 2, "WEATHER"),
    (CONTROL_NUM + 3, "WEATHER"),
    (CONTROL_NUM + 4, LOCAL_HAZARD),
    (CONTROL_NUM + 5, "WEATHER"),
    (CONTROL_NUM + 6, "TRAFFIC")
]

# expand burst (required)
for i in range(CONTROL_NUM + 2):
    burst.append((CONTROL_NUM + 7 + i, f"CUSTOM_{i}"))

# tracking flags
weather_at_c3 = "N"
local_hazard_appended = "N"

# sliding window processing
for alert in burst:

    categories = [x[1] for x in dq]
    counts = Counter(categories)

    # spam rule: reject if already 2 or more
    if counts[alert[1]] >= 2:
        continue

    dq.append(alert)

    # check required conditions
    if alert[1] == "WEATHER" and alert[0] == CONTROL_NUM + 3:
        weather_at_c3 = "Y"

    if alert[1] == LOCAL_HAZARD:
        local_hazard_appended = "Y"

# final computations
timestamp_sum = sum(x[0] for x in dq)

# ================================
# OUTPUTS (ASSESSMENT TABLE)
# ================================

print("LOCAL HAZARD USED:", LOCAL_HAZARD)
print("TOTAL BURST SIZE:", len(burst))
print("WEATHER AT C+3 APPENDED:", weather_at_c3)
print("LOCAL HAZARD APPENDED:", local_hazard_appended)
print("FINAL DEQUE LENGTH:", len(dq))
print("FINAL TIMESTAMP SUM:", timestamp_sum)
print("FINAL DEQUE OUTPUT:", dq)
```
